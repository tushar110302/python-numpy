```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1, 2, 3], [4, 5, 7]])
array2 = np.random.random((3, 3))
array3 = np.zeros((4, 4))

print(array1)
print(array2)
print(array3)
```

```python
np.save("array1.npy", array1)
np.save("array2.npy", array2)
np.save("array3.npy", array3)
```

```python
loaded_arr1 = np.load("array1.npy")
loaded_arr2 = np.load("array2.npy")
loaded_arr3 = np.load("array3.npy")

print(loaded_arr1)
print(loaded_arr2)
print(loaded_arr3)
```

```python
try:
    loaded_img = np.load("numpy-logo.npy")

    plt.figure(figsize=(10,5))
    plt.subplot(121)
    plt.imshow(loaded_img)
    plt.title("Numpy Logo (Light)")

    dark_logo = 1 - loaded_img
    plt.subplot(122)
    plt.imshow(dark_logo)
    plt.title("Numpy Logo (Dark)")

except FileNotFoundError:
    print("Numpy logo title not found")
```
